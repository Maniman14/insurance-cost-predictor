```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv('insurance.csv')
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.sex.value_counts()
```

```python
data.region.value_counts()
```

```python
encoder = LabelEncoder()
```

```python
labels = encoder.fit_transform(data.sex)
```

```python
data['sex'] = labels
```

```python
data.head()
```

```python
labels = encoder.fit_transform(data.smoker)
```

```python
data['smoker'] = labels
```

```python
data.head()
```

```python
labels = encoder.fit_transform(data.region)
```

```python
data['region'] = labels
```

```python
data.head()
```

```python
X = data.drop(columns='charges')
Y = data['charges']

print(X.head())
```

```python
print(Y.head())
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=101)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
model = RandomForestRegressor()
```

```python
model.fit(X_train,Y_train)
```

```python
testing_data_prediction = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

mse = mean_squared_error(Y_test, testing_data_prediction)
mae = mean_absolute_error(Y_test, testing_data_prediction)
r2 = r2_score(Y_test, testing_data_prediction)

print(f"MSE: {mse:.2f}")
print(f"MAE: {mae:.2f}")
print(f"R² Score: {r2:.2f}")
```

```python
def predict_cost(age, sex, bmi, children, smoker, region):
    input_data = pd.DataFrame([{
        'age': age,
        'sex': sex,
        'bmi': bmi,
        'children': children,
        'smoker': smoker,
        'region': region
    }])
    return model.predict(input_data)[0]
```
